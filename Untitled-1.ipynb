```python
G = 6.67430e-11 # Constante de gravitaciÃƒÂ³n universal en m^3 kg^-1 s^-2
M_sol = 1.989e30 # Masa del Sol en kg
M_tierra = 5.972e24 # Masa de la Tierra en kg
M_apophis = 2.7e10 # Masa de Apophis en kg
M_jupiter = 1.898e27 # Masa de Jupiter en kg
M_luna = 7.342e22 # Masa de la Luna en kg
M_marte = 6.417e23 # Masa de Marte en kg
M_mercurio = 3.301e23 # Masa de Mercurio en kg
M_venus = 4.867e24 # Masa de Venus en kg
M_saturno = 5.683e26 # Masa de Saturno en kg
M_urano = 8.681e25 # Masa de Urano en kg
M_neptuno = 1.024e26 # Masa de Neptuno en kg
M_ceres = 9.393e20 # Masa de Ceres en kg
M_pallas = 2.14e20 # Masa de Pallas en kg
M_vesta = 2.59e20 # Masa de Vesta en kg
M_hygiea = 8.32e19 # Masa de Hygiea en kg
```

```python

# Tiempo inicial y final en JD
tiempo_inicio = Time(fecha_inicio, scale='tdb').jd
tiempo_final  = Time(fecha_final, scale='tdb').jd

# Duración total
tiempo_simulacion_dias = tiempo_final - tiempo_inicio

# Resolución temporal
paso_en_dias = 1.0 / 24.0   # 1 hora

# Número de pasos
num_pasos = int(tiempo_simulacion_dias / paso_en_dias) + 1

# Vector temporal
t_jd = np.linspace(tiempo_inicio,
                   tiempo_final,
                   num_pasos)

# Tiempo relativo
t_dias = t_jd - tiempo_inicio

# Conversión a años calendario
t_anios = t_dias / 365.25

anio_inicio = Time(fecha_inicio,
                   scale='tdb').decimalyear

anios_calendario = anio_inicio + t_anios

print(f"Pasos totales: {num_pasos}")
print(f"Resolución temporal: {t_jd[1]-t_jd[0]:.5f} días")
```

```python
# tiempo en días
t_dias = np.linspace(0, tiempo_simulacion_dias, num_pasos)

# convertir a segundos
t_seg = t_dias * 86400

# convertir a unidades canónicas
ts = t_seg / U_T
```

```python
paso_muestreo = 1  # días (entero)
t_dias_rb = t_dias[::paso_muestreo]
t_anios_rb = t_dias_rb / 365.25
anios_rb = anio_inicio + t_anios_rb
t_seg_rb = t_dias_rb * 86400.0
t_can_rb = t_seg_rb / U_T
def simular_y_analizar_rebound(sistema, nombres, colores, i_tierra, i_apophis, fecha_inicio, t_dias_rb, U_L, U_T, titulo="REBOUND"):
    sim = rb.Simulation()
    sim.G = 1.0
    sim.integrator = 'ias15'

    # Se añaden dinámicamente todos los cuerpos del sistema recibido
    for body in sistema:
        r = body['r']
        v = body['v']
        sim.add(m=body['m'], x=r[0], y=r[1], z=r[2], vx=v[0], vy=v[1], vz=v[2]) 

    # Configuración de tiempos
    t_seg_rb = t_dias_rb * 86400.0
    t_can_rb = t_seg_rb / U_T
    anios_rb = Time(fecha_inicio).decimalyear + t_dias_rb / 365.25

    # Reservar arreglos para posiciones
    rs_rb = np.zeros((len(sistema), len(t_can_rb), 3))

    print(f"[{titulo}] Integrando {len(sistema)} cuerpos...")
    for k, t in enumerate(t_can_rb):
        sim.integrate(float(t))
        for i, p in enumerate(sim.particles):
            rs_rb[i, k, 0] = p.x
            rs_rb[i, k, 1] = p.y
            rs_rb[i, k, 2] = p.z

    # Se obtienen coordenadas heliocéntricas respecto al cuerpo 0 (Sol)
    rs_rb_helio = rs_rb - rs_rb[0:1, :, :]

    # =====================================================
    # Gráficos Generales: Órbitas y Distancia
    # =====================================================
    fig, ax = plt.subplots(1, 2, figsize=(17, 5))

    ax[0].scatter(0.0, 0.0, color='gold', s=80, label='Sol', zorder=5)
    
    # Graficar todos los cuerpos recibidos excepto el Sol index 0
    for idx in range(1, len(sistema)):
        # nos aseguramos que haya color para el indice
        color = colores[idx] if idx < len(colores) else 'gray'
        ax[0].plot(rs_rb_helio[idx, :, 0], rs_rb_helio[idx, :, 1], color=color, lw=1.3, label=nombres[idx])
        
    ax[0].set_title(f'Órbitas heliocéntricas (XY) - {titulo}')
    ax[0].set_xlabel('X (UA)')
    ax[0].set_ylabel('Y (UA)')
    ax[0].axis('equal')
    ax[0].grid(True, alpha=0.3)
    ax[0].legend(fontsize=8, loc='upper right')

    dist_rb_ua = np.linalg.norm(rs_rb_helio[i_apophis] - rs_rb_helio[i_tierra], axis=1)
    dist_rb_km = dist_rb_ua * U_L / 1e3
    
    ax[1].plot(anios_rb, dist_rb_km, color='black', lw=1.2)
    ax[1].set_title(f'Distancia Apophis-Tierra - {titulo}')
    ax[1].set_xlabel('Año calendario')
    ax[1].set_ylabel('Distancia (km)')
    ax[1].grid(True, alpha=0.3)

    # Identificar acercamiento mínimo
    idx_min = np.argmin(dist_rb_km)
    d_min_km = dist_rb_km[idx_min]
    d_min_ua = dist_rb_ua[idx_min]
    ax[1].scatter(anios_rb[idx_min], d_min_km, color='red', s=50, label='Min Distancia')
    ax[1].legend()

    plt.tight_layout()
    plt.show()

    # =====================================================
    # Análisis matemático para el evento
    # =====================================================
    t_dias_min = t_dias_rb[idx_min]
    fecha_min = Time(fecha_inicio).to_datetime() + np.timedelta64(int(t_dias_min), 'D')
    fecha_referencia = Time('2029-04-13').to_datetime()
    delta_dias_ref = (fecha_min - fecha_referencia).days

    print('='*60)
    print(f'RESUMEN: DISTANCIA MINIMA TIERRA-APOPHIS ({titulo})')
    print('='*60)
    print(f'Fecha del acercamiento: {fecha_min:%Y-%m-%d}')
    print(f'Distancia minima: {d_min_km:>12,.1f} km')
    print(f'Distancia minima: {d_min_ua:>12.6f} UA')
    print(f'Fecha referencia (2029-04-13): {fecha_referencia:%Y-%m-%d}')
    print(f'Diferencia: {delta_dias_ref:>+4d} dias')
    print('='*60)

    # =====================================================
    # Gráficos en Zoom de la trayectoria de Apophis vs Tierra
    # =====================================================
    ancho = 150
    i_start = max(0, idx_min - ancho)
    i_end = min(len(t_dias_rb), idx_min + ancho + 1)

    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

    ax1.plot(anios_rb[i_start:i_end], dist_rb_km[i_start:i_end], 'b-', lw=2)
    ax1.scatter([anios_rb[idx_min]], [d_min_km], color='red', s=100, zorder=5)
    ax1.axvline(anios_rb[idx_min], color='red', linestyle='--', alpha=0.5)
    ax1.set_xlabel('Año calendario')
    ax1.set_ylabel('Distancia (km)')
    ax1.set_title(f'Acercamiento Tierra-Apophis (zoom) - {titulo}')
    ax1.grid(True, alpha=0.3)

    r_tierra_zoom = rs_rb_helio[i_tierra, i_start:i_end] / U_L
    r_apophis_zoom = rs_rb_helio[i_apophis, i_start:i_end] / U_L

    ax2.plot(r_tierra_zoom[:, 0], r_tierra_zoom[:, 1], 'b-', lw=2, label='Tierra')
    ax2.plot(r_apophis_zoom[:, 0], r_apophis_zoom[:, 1], 'r-', lw=2, label='Apophis')
    ax2.scatter([r_tierra_zoom[idx_min - i_start, 0]], [r_tierra_zoom[idx_min - i_start, 1]], 
                color='blue', s=100, marker='o', zorder=5)
    ax2.scatter([r_apophis_zoom[idx_min - i_start, 0]], [r_apophis_zoom[idx_min - i_start, 1]], 
                color='red', s=100, marker='o', zorder=5)
    ax2.set_xlabel('X (UA)')
    ax2.set_ylabel('Y (UA)')
    ax2.set_title(f'Orbitas heliocentricas (zoom) - {titulo}')
    ax2.grid(True, alpha=0.3)
    ax2.legend()
    ax2.axis('equal')

    plt.tight_layout()
    plt.show()

    return rs_rb_helio, d_min_km, fecha_min
```

```python
# ============================================================
# EXTRACCIÓN DE MASAS Y POSICIONES
# ============================================================

m_S = sistema_2_cuerpos[0]['m']
r_S_orig = np.array(sistema_2_cuerpos[0]['r'])
v_S_orig = np.array(sistema_2_cuerpos[0]['v'])

m_T = sistema_2_cuerpos[1]['m']
r_T_orig = np.array(sistema_2_cuerpos[1]['r'])
v_T_orig = np.array(sistema_2_cuerpos[1]['v'])

m_L = sistema_2_cuerpos[2]['m']
r_L_orig = np.array(sistema_2_cuerpos[2]['r'])
v_L_orig = np.array(sistema_2_cuerpos[2]['v'])

m_A = sistema_2_cuerpos[3]['m']
r_A_orig = np.array(sistema_2_cuerpos[3]['r'])
v_A_orig = np.array(sistema_2_cuerpos[3]['v'])

# ============================================================
# CONSTRUCCIÓN DE VECTORES RELATIVOS HACIA ARRIBA
# ============================================================

m_TL = m_T + m_L
m_TLA = m_A + m_TL

# Centros de masa intermedios
r_TL_cm = (m_T * r_T_orig + m_L * r_L_orig) / m_TL
v_TL_cm = (m_T * v_T_orig + m_L * v_L_orig) / m_TL

r_TLA_cm = (m_A * r_A_orig + m_TL * r_TL_cm) / m_TLA
v_TLA_cm = (m_A * v_A_orig + m_TL * v_TL_cm) / m_TLA

# Vectores relativos
r_F_TL = r_T_orig - r_L_orig
v_F_TL = v_T_orig - v_L_orig

r_F_ATL = r_A_orig - r_TL_cm
v_F_ATL = v_A_orig - v_TL_cm

r_S_TLA = r_S_orig - r_TLA_cm
v_S_TLA = v_S_orig - v_TLA_cm

# Centro de masa global (origen)
RCM_STLA = np.zeros(3)
VCM_STLA = np.zeros(3)

# ============================================================
# DESCOMPOSICIÓN JERÁRQUICA COMPLETA PARA 4 PARTÍCULAS: S, T, L, A
# ============================================================

# 1) Sistema S - TLA (r_S_TLA = r_S - r_TLA)
r_S = RCM_STLA + (m_TLA / (m_S + m_TLA)) * r_S_TLA
v_S = VCM_STLA + (m_TLA / (m_S + m_TLA)) * v_S_TLA

RCM_TLA = RCM_STLA - (m_S / (m_S + m_TLA)) * r_S_TLA
VCM_TLA = VCM_STLA - (m_S / (m_S + m_TLA)) * v_S_TLA

# 2) Sistema A - TL (r_F_ATL = r_A - r_TL)
r_A = RCM_TLA + (m_TL / (m_A + m_TL)) * r_F_ATL
v_A = VCM_TLA + (m_TL / (m_A + m_TL)) * v_F_ATL

RCM_TL = RCM_TLA - (m_A / (m_A + m_TL)) * r_F_ATL
VCM_TL = VCM_TLA - (m_A / (m_A + m_TL)) * v_F_ATL

# 3) Sistema T - L (r_F_TL = r_T - r_L)
r_T = RCM_TL + (m_L / (m_T + m_L)) * r_F_TL
v_T = VCM_TL + (m_L / (m_T + m_L)) * v_F_TL

r_L = RCM_TL - (m_T / (m_T + m_L)) * r_F_TL
v_L = VCM_TL - (m_T / (m_T + m_L)) * v_F_TL


# ============================================================
# SISTEMA FINAL
# ============================================================

sistema_2_jerarquico_nuevo = [
    dict(m=m_S, r=r_S, v=v_S),
    dict(m=m_T, r=r_T, v=v_T),
    dict(m=m_L, r=r_L, v=v_L),
    dict(m=m_A, r=r_A, v=v_A)
]
```

```python
def calcular_elementos_orbitales(rvec, vvec, mu):
    r = np.linalg.norm(rvec)
    v = np.linalg.norm(vvec)

    # Momento angular
    hvec = np.cross(rvec, vvec)
    h = np.linalg.norm(hvec)

    # Vector de excentricidad (Laplace)
    evec = np.cross(vvec, hvec) / mu - rvec / r
    e = np.linalg.norm(evec)

    # Vector de nodos
    nvec = np.cross([0, 0, 1], hvec)
    n = np.linalg.norm(nvec)

    # Elementos orbitales
    p = h**2 / mu
    I = np.arccos(hvec[2] / h)

    # Ascensión recta del nodo ascendente (Omega)
    if n == 0:
        Omega = 0 # Órbita ecuatorial, nodo indefinido
    else:
        Omegap = np.arccos(np.dot(nvec, [1, 0, 0]) / n)
        if nvec[1] > 0:
            Omega = Omegap
        else:
            Omega = 2 * np.pi - Omegap

    # Argumento del periapsis (omega)
    if e == 0:
        omegap = 0 # Órbita circular, periapsis indefinida
    elif n == 0:
        omegap = np.arctan2(evec[1], evec[0]) # Órbita ecuatorial, nodo indefinido
    else:
        omegap = np.arccos(np.dot(nvec, evec) / (n * e))
        if evec[2] < 0:
            omegap = 2 * np.pi - omegap

    # Anomalía verdadera (f)
    fp = np.arccos(np.dot(rvec, evec) / (r * e))
    if np.dot(rvec, vvec) > 0:
        f = fp
    else:
        f = 2 * np.pi - fp

    return p, e, I, Omega, omegap, f, hvec, nvec
```

```python
p_apophis, e_apophis, I_apophis, Omega_apophis, omegap_apophis, f_apophis, hvec, nvec = calcular_elementos_orbitales(r_vec_apophis, v_vec_apophis, mu)
q_apophis = p_apophis / (1 + e_apophis)
```
