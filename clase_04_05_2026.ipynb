```python
import pymcel as pc
import numpy as np
import matplotlib.pyplot as plt
```

```python
sistema = [
    dict(m=100, r=[-0.142857, 0, 0], v=[0, -2.958, 0]),
    dict(m=5, r=[2.85714, 0, 0], v=[0, 2.958, 0]),
    dict(m=0.01, r=[3.62345, 0, 0], v=[0, 4.9700107, 0])
    #dict(m=0.01, r=[3.63345, 0, 0], v=[0, 4.9700107, 0])
]
ts = np.linspace(0, 8, 800)
# Agrega 'constantes' aquí:
rs, vs, rps, vps, constantes = pc.ncuerpos_solucion(sistema, ts)
```

```python
plt.plot(rps[0][:,0], rps[0][:,1], label="cuerpo 1")
plt.plot(rps[1][:,0], rps[1][:,1], label="cuerpo 2")
plt.plot(rps[2][:,0], rps[2][:,1], label="cuerpo 3")
plt.axis("equal")
plt.legend()
```

```python
from celluloid import Camera
```

```python
fig = plt.figure()
camera = Camera(fig)
for i in range(len(ts)):
    plt.plot(rps[0,i,0], rps[0,i,1], 'ko', label="cuerpo 1")
    plt.plot(rps[1,i,0], rps[1,i,1], 'ro', label="cuerpo 2")
    plt.plot(rps[2,i,0], rps[2,i,1], 'bo', label="cuerpo 3")
    camera.snap()
plt.axis("equal")
    
 
```
