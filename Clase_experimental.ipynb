```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import poisson

# 1. Ingresa aquí tus datos ACUMULATIVOS reales
# (Ejemplo de conteo acumulativo: empieza en 0 y va subiendo)
datos_acumulativos = [0, 2, 5, 8, 12, 14, 19, 22, 26, 27, 30, 32, 36, 39, 44, 46, 49, 53, 56, 58]

# 2. Calcular los conteos por intervalo
# np.diff resta cada valor con el anterior (datos[i+1] - datos[i])
conteos_por_intervalo = np.diff(datos_acumulativos)
print(f"Conteos individuales de muones: {conteos_por_intervalo}")

# 3. Calcular el parámetro lambda (λ) - la media de los intervalos
lambda_param = np.mean(conteos_por_intervalo)
print(f"Parámetro lambda (media) calculado: {lambda_param:.3f}")

# 4. Generar los valores para la distribución teórica de Poisson
x = np.arange(0, max(conteos_por_intervalo) + 3)
pmf_poisson = poisson.pmf(x, mu=lambda_param)

# 5. Crear la gráfica
plt.figure(figsize=(8, 5))

# Histograma de los conteos individuales
plt.hist(conteos_por_intervalo, bins=np.arange(0, max(conteos_por_intervalo) + 2) - 0.5, density=True, 
         alpha=0.6, color='skyblue', edgecolor='black', label='Histograma (Datos reales)')

# Curva teórica de Poisson
plt.plot(x, pmf_poisson, 'ro-', ms=8, label=f'Distribución Poisson ($\lambda = {lambda_param:.2f}$)')
plt.vlines(x, 0, pmf_poisson, colors='r', lw=2, alpha=0.5)

# Detalles de la gráfica
plt.xlabel('Número de muones por intervalo ($k$)')
plt.ylabel('Probabilidad (Densidad)')
plt.title('Ajuste de distribución de Poisson para conteo de muones')
plt.xticks(x)
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.show()
```
